```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('car_fuel_efficiency.csv')
df.head()
```

```python
pandas_version = pd.__version__
print(f'The pandas version is {pandas_version}')
```

```python
Records_count = df.shape[0]
print(f'The records count is {Records_count}')
```

```python
fuel_types = df['fuel_type'].nunique()
print(f'There are {fuel_types} fuel types in the data')
```

```python
missing_values = 0
for col in df.columns:
    if df[col].isnull().sum() != 0:
        missing_values+=1
    else:
        continue
print(f'The number of columns with missing values is {missing_values}')
```

```python
max_fuel_efficiency = df[df['origin'] == 'Asia']['fuel_efficiency_mpg'].max().round(3)
print(f'The maximum fuel efficiency is {max_fuel_efficiency}')
```

```python
median_original = df['horsepower'].median()
print(f"Original median horsepower: {median_original}")
mode_value = df['horsepower'].mode()[0]
print(f"Most frequent horsepower value (mode): {mode_value}")
df_filled = df.copy()
df_filled['horsepower'] = df_filled['horsepower'].fillna(mode_value)
print(f"Missing values after filling: {df_filled['horsepower'].isna().sum()}")
median_after_fill = df_filled['horsepower'].median()
print(f"Median horsepower after filling: {median_after_fill}")
# Compare the medians
print("\n" + "="*50)
print("COMPARISON:")
print(f"Original median: {median_original}")
print(f"After filling median: {median_after_fill}")
print(f"Difference: {median_after_fill - median_original}")

if median_after_fill > median_original:
    print("\nAnswer: Yes, it increased")
elif median_after_fill < median_original:
    print("\nAnswer: Yes, it decreased")
else:
    print("\nAnswer: No, it stayed the same")
```

```python
asian_cars = df[df['origin'] == 'Asia']
print(f"Total Asian cars: {len(asian_cars)}")

# Step 2: Select only columns vehicle_weight and model_year
# Step 3: Select the first 7 values
selected_data = asian_cars[['vehicle_weight', 'model_year']].head(7)
print("\nFirst 7 Asian cars:")
print(selected_data)
X = selected_data.values
print(f"\nX shape: {X.shape}")
print("X matrix:")
print(X)
XTX = X.T @ X
print("\nXTX (X^T * X):")
print(XTX)
XTX_inv = np.linalg.inv(XTX)
print("\nInverse of XTX:")
print(XTX_inv)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
print("\ny:")
print(y)
w = XTX_inv @ X.T @ y
print("\nw (result):")
print(w)
result_sum = w.sum()
print(f"\nSum of all elements in w: {result_sum}")
```
